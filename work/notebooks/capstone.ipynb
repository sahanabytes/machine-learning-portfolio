Abstract
This study examines how rule-based scoring and machine learning can support website content refresh prioritization. Using a 30,000-row anonymized content dataset covering 32 clients, the study compares a Week 4 rule-based baseline with a Decision Tree classifier. The model was evaluated using a grouped train-test split to keep clients separate and was measured using Precision@20 and Precision@50. ## 2. Data

### 2. Data

This study uses the anonymized content refresh dataset provided in the FlyRank ML Internship repository. The modelling dataset, `content_refresh_anonymized.csv`, contains 30,000 rows and 44 columns, representing content performance information across 32 anonymized clients. The dataset is a small, public-safe starter sample and is the dataset used for the final modelling experiment.

The wider FlyRank internship warehouse contains approximately 79 million rows. Its daily performance data is stored in the `fact_content_daily_performance` table in monthly Parquet files. In the Week 3 data-contract experiment, the March 2026 partition contained 9,841,378 rows, covering March 1–31, 2026. A separate March-to-April analysis matched 158,549 client-content pairs with available Google Search Console data in both months. This warehouse experiment is documented as a separate data-contract and label investigation; it should not be confused with the 30,000-row dataset used to train the final Decision Tree.

The final modelling dataset includes trailing 90-day performance measures, including impressions, search volume, average position and click-through rate, along with content attributes such as content age, word count and days since the last update. The target label was defined as 1 when `trend_direction` was `down`, and 0 otherwise.

Client and content identifiers were retained only for grouping, traceability and analysis, not as predictive features. Trend-derived information was used to construct the target and was excluded from the model features to avoid target leakage. Missing feature values were handled using medians calculated from the training data, with a fallback of zero where needed. No client names, private URLs, page titles or search queries are included in this public report.

## 3. Methodology

*Assumptions, features, label definition, baseline, validation design, leakage checks.*

### 3. Methodology

This study compares a simple rule-based baseline with a Decision Tree classifier for identifying content pages with observed declining trends.

The target variable, `decline_label`, was created from the `trend_direction` field. Pages labelled `down` were assigned 1, while all other trend categories were assigned 0. This means the model identifies observed declining pages; it does not predict whether a content refresh will succeed.

Eight numerical features were used: `days_since_last_update`, `impressions_90d`, `search_volume`, `competition`, `word_count`, `content_age_days`, `avg_position` and `ctr`. Client identifiers were used only for grouping, not as model features. The target and trend-derived fields were excluded from the feature set.

The data was divided using a grouped 75:25 train-test split. The split kept all rows belonging to each client in only one set. This produced 22,885 training rows from 24 clients and 7,115 test rows from 8 different clients, with no client overlap.

Missing numerical feature values were filled using medians calculated from the training set. A Decision Tree classifier with a maximum depth of 3 and a minimum of 100 samples per leaf was trained on the training data.

The Week 4 rule-based baseline and the Decision Tree were evaluated on the same held-out test set. Precision@20 and Precision@50 measured the proportion of declining pages among the highest-ranked 20 and 50 pages. A leakage audit confirmed that the target was not included in the model features.

## 4. Results (vs baseline)

*Model vs baseline on the same split. The honest table.*

### 4. Results: Model vs Baseline

The Decision Tree was compared with the Week 4 rule-based baseline using the same held-out test set of 7,115 rows from 8 unseen clients.

| Method | Precision@20 | Precision@50 |
|---|---:|---:|
| Week 4 baseline | 0.60 | 0.60 |
| Decision Tree | 0.45 | 0.56 |

The Week 4 baseline performed better than the Decision Tree at both ranking cutoffs. Among the top 20 pages selected by the baseline, 60% were labelled as declining, compared with 45% for the Decision Tree. At the top 50, the baseline achieved 60% precision, while the Decision Tree achieved 56%.

These results show that the more complex model did not outperform the simpler rule-based approach in this experiment. The baseline is therefore the stronger method for prioritizing pages for human review based on this held-out evaluation. The results are specific to this dataset and split and should be treated as directional rather than proof of performance on all websites.

## 5. Limitations

*What this work cannot claim.*

### 5. Limitations

This study has several limitations.

1. **Limited dataset:** The final model was trained using a 30,000-row anonymized sample containing data from 32 clients. The results may not represent all websites or content types.

2. **Limited test coverage:** The grouped test set contained only 8 clients. Although this prevents client overlap between training and testing, results from a small number of held-out clients may not generalize to every client.

3. **Observed decline is not future refresh success:** The target label represents pages with an observed `down` trend. It does not show whether updating those pages would improve their performance.

4. **Baseline outperformed the model:** The Decision Tree achieved lower Precision@20 and Precision@50 than the Week 4 rule-based baseline. Therefore, the experiment does not demonstrate that machine learning is better than the simpler method.

5. **Prediction errors:** The Decision Tree produced 2,664 false positives and 433 false negatives. This means it incorrectly flagged some non-declining pages and missed some pages labelled as declining.

6. **Feature limitations:** The model relied mainly on impressions, content age, average position and click-through rate. Other features had zero importance in this fitted tree, so their usefulness was not demonstrated in this experiment.

7. **No causal evidence:** The analysis identifies patterns associated with observed decline. It does not establish that content age, impressions or content updates directly cause performance changes.

The findings should therefore be treated as directional decision-support evidence. Human review and further testing are necessary before using the approach in a real content-refresh workflow.

## 6. Ranked recommendations

*The action playbook output — the paper's recommendations section.*

### 6. Ranked Recommendations

Based on the experimental results, the following actions are recommended for content teams.

**1. Use the Week 4 baseline as the current prioritization method.**

The rule-based baseline achieved Precision@20 of 0.60 and Precision@50 of 0.60, outperforming the Decision Tree on both measures. It is the stronger approach in this experiment and should be preferred for an initial review workflow.

**2. Prioritize older content with sufficient search visibility.**

The baseline assigns 50 points when `days_since_last_update` is at least 180 and another 50 points when `impressions_90d` is at least 500. Pages meeting both conditions receive the highest score of 100 and can be considered for earlier human review.

**3. Keep human review in the workflow.**

A high score should identify a page for investigation, not trigger an automatic content update. Reviewers should examine the page's relevance, accuracy, search intent and available performance data before deciding what action to take.

**4. Investigate pages flagged by the model but not by the baseline.**

The Decision Tree may provide additional signals for further investigation, even though it did not outperform the baseline. These pages can be reviewed as a secondary group rather than treated as confirmed decline cases.

**5. Monitor outcomes after content changes.**

Content teams should track performance after updates and compare results over an appropriate period. This would help determine whether the prioritization process is useful and whether refreshed pages show measurable improvement.

**Overall recommendation:** Start with the simpler Week 4 baseline, retain human judgment, and collect further evidence before adopting the Decision Tree as the main prioritization method.
